```python
# ============================================================
# 01. EXPLORATORY DATA ANALYSIS (EDA)
# AML Transaction Monitoring System
# ============================================================

#import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


# ============================================================
# GOOGLE DRIVE
# ============================================================

#drive.mount('/content/drive')

#print("Google Drive mounted successfully.")


# EXTRACT DATASET

#zip_path = "/content/drive/MyDrive/AML_Project/archive.zip"
#extract_path = "/content/aml_data"

#os.makedirs(extract_path, exist_ok=True)

#print("\nExtracting dataset...")

#os.system(
    #'unzip -o "/content/drive/MyDrive/AML_Project/archive.zip" '
    #'-d "/content/aml_data"'
#)

#print("Extraction completed.")


# ============================================================
# LOAD DATASET
# ============================================================

file_path = "../data/raw/SAML-D.csv"

print("\nLoading dataset...")

df = pd.read_csv(file_path)

print("Dataset loaded successfully.")
print("Shape:", df.shape)


# ============================================================
# BASIC DATASET INFORMATION
# ============================================================

print("\n================================================")
print("BASIC DATASET INFORMATION")
print("================================================")

print("Number of rows    :", df.shape[0])
print("Number of columns :", df.shape[1])

print("\nColumns:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)


# ============================================================
# FIRST FIVE ROWS
# ============================================================

print("\n================================================")
print("FIRST FIVE ROWS")
print("================================================")

display(df.head())


# ============================================================
# MISSING VALUES
# ============================================================

print("\n================================================")
print("MISSING VALUES")
print("================================================")

missing_values = df.isnull().sum()

print(missing_values)

print("\nTotal missing values:", missing_values.sum())


# ============================================================
# DUPLICATE ROWS
# ============================================================

print("\n================================================")
print("DUPLICATE ROWS")
print("================================================")

duplicate_count = df.duplicated().sum()

print("Duplicate rows:", duplicate_count)


# ============================================================
# TARGET VARIABLE ANALYSIS
# ============================================================

print("\n================================================")
print("TARGET VARIABLE ANALYSIS")
print("================================================")

target_counts = df["Is_laundering"].value_counts()

print(target_counts)

print("\nTarget percentages:")

target_percentage = (
    df["Is_laundering"].value_counts(normalize=True) * 100
)

print(target_percentage)


# Target distribution plot

plt.figure(figsize=(6, 4))

target_counts.plot(kind="bar")

plt.title("Laundering vs Legitimate Transactions")
plt.xlabel("Is Laundering")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=0)

plt.show()


# ============================================================
# TRANSACTION AMOUNT ANALYSIS
# ============================================================

print("\n================================================")
print("TRANSACTION AMOUNT ANALYSIS")
print("================================================")

print(df["Amount"].describe())

print("\nMinimum amount:", df["Amount"].min())
print("Maximum amount:", df["Amount"].max())


# Use a sample for visualization to reduce memory usage

sample_size = min(200000, len(df))

eda_sample = df.sample(
    n=sample_size,
    random_state=42
)

plt.figure(figsize=(8, 5))

plt.hist(
    eda_sample["Amount"],
    bins=100
)

plt.title("Transaction Amount Distribution")
plt.xlabel("Amount")
plt.ylabel("Frequency")

plt.show()


# ============================================================
# PAYMENT TYPE ANALYSIS
# ============================================================

print("\n================================================")
print("PAYMENT TYPE ANALYSIS")
print("================================================")

payment_table = pd.crosstab(
    df["Payment_type"],
    df["Is_laundering"]
)

print(payment_table)

payment_table.plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Payment Type vs Laundering")
plt.xlabel("Payment Type")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=45)

plt.show()


# ============================================================
# PAYMENT CURRENCY ANALYSIS
# ============================================================

print("\n================================================")
print("PAYMENT CURRENCY ANALYSIS")
print("================================================")

currency_table = pd.crosstab(
    df["Payment_currency"],
    df["Is_laundering"]
)

print(currency_table)

currency_table.plot(
    kind="bar",
    figsize=(12, 5)
)

plt.title("Payment Currency vs Laundering")
plt.xlabel("Payment Currency")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=45)

plt.show()


# ============================================================
# SENDER BANK LOCATION ANALYSIS
# ============================================================

print("\n================================================")
print("SENDER BANK LOCATION ANALYSIS")
print("================================================")

sender_location_table = pd.crosstab(
    df["Sender_bank_location"],
    df["Is_laundering"]
)

print(sender_location_table)

sender_location_table.plot(
    kind="bar",
    figsize=(12, 5)
)

plt.title("Sender Bank Location vs Laundering")
plt.xlabel("Sender Bank Location")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=45)

plt.show()


# ============================================================
# RECEIVER BANK LOCATION ANALYSIS
# ============================================================

print("\n================================================")
print("RECEIVER BANK LOCATION ANALYSIS")
print("================================================")

receiver_location_table = pd.crosstab(
    df["Receiver_bank_location"],
    df["Is_laundering"]
)

print(receiver_location_table)

receiver_location_table.plot(
    kind="bar",
    figsize=(12, 5)
)

plt.title("Receiver Bank Location vs Laundering")
plt.xlabel("Receiver Bank Location")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=45)

plt.show()


# ============================================================
# LAUNDERING TYPE ANALYSIS
# ============================================================

print("\n================================================")
print("LAUNDERING TYPE ANALYSIS")
print("================================================")

print(
    df["Laundering_type"].value_counts(
        dropna=False
    )
)


# ============================================================
# EDA SUMMARY
# ============================================================

print("\n================================================")
print("EDA COMPLETED")
print("================================================")

print("\nDataset shape:", df.shape)

print(
    "Total missing values:",
    df.isnull().sum().sum()
)

print(
    "Duplicate rows:",
    df.duplicated().sum()
)

print(
    "Laundering transactions:",
    (df["Is_laundering"] == 1).sum()
)

print(
    "Legitimate transactions:",
    (df["Is_laundering"] == 0).sum()
)

print("\n================================================")
print("EXPLORATORY DATA ANALYSIS FINISHED")
print("================================================")
```
